Task: 

    Create a graph where you pass in a single list of integers along with a name and an operation. If the operation is a " + ", you add the elements, and if it is a " * ", you multiply the elements, all within the same node.

Input: 
    { name: "Jack Sparrow", values: [1 , 2 ,3 , 4] ,operation: " * "} 

Output: Hi Jack Sparrow, your answer is 24.

```python
from typing import TypedDict , List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    values:List[int]
    operation: str
    result: str
    
```

```python
from functools import reduce

def perform_operation(state:AgentState)->AgentState:
    if (state["operation"] == "+"):
        state["result"] = f"hey {state['name']} , your answer is {sum(state['values'])}"
    elif (state["operation"] == "*"):
        state["result"] = f"hey {state['name']} , your answer is {reduce(lambda x,y:x*y,state['values'])}"
   
    return state
```

```python
graph  = StateGraph(AgentState)

graph.add_node("Math_Operation",perform_operation)

graph.set_entry_point("Math_Operation")
graph.set_finish_point("Math_Operation")

app = graph.compile()
```

```python
result = app.invoke({
    "name":"john",
    "values":[1,2,3,4],
    "operation":"+"
    })
```
