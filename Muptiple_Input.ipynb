Objectives: 
1. Define a more complex AgentState. 
2. Create a processing node that performs operation on list data. 
3. Set up a LangGraph that processes and outputs computed result. 
4. Invoke the graph with structured inputs and retrieve outputs. 

Main goal: Learn how to handle multiple inputs.

```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values : list[int]
    name : str
    result : str
```

```python
def process_values(state : AgentState) -> AgentState:
    """ This function handles multiple different inputs"""
    print(state)
    state["result"] = f"Hi there , {state['name']} ! Your sum = {sum(state['values'])}"
    print(state)
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("process_values",process_values)

graph.set_entry_point("process_values")
graph.set_finish_point("process_values")

app = graph.compile()


```

```python
result = app.invoke({"name":"John","values":[1,2,3,4,5]})
result
```
